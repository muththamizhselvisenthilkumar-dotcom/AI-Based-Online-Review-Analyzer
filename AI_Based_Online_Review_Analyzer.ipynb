```python
import nltk
import re

from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()

print("AI-Based Online Review Analyzer")
print("System is ready!")
```

```python
review = input("Enter your online review: ")

print("\nYour Review:")
print(review)
```

```python
# Sentiment analysis
score = sia.polarity_scores(review)

compound = score["compound"]

if compound >= 0.05:
    sentiment = "Positive"

elif compound <= -0.05:
    sentiment = "Negative"

else:
    sentiment = "Neutral"


# Word analysis
words = re.findall(r'\b[a-zA-Z]+\b', review.lower())

word_count = len(words)


# Important keywords
keywords = [
    "quality",
    "price",
    "delivery",
    "product",
    "service",
    "quality",
    "design",
    "customer",
    "support",
    "excellent",
    "bad",
    "good"
]

found_keywords = []

for word in keywords:
    if word in words and word not in found_keywords:
        found_keywords.append(word)


# Recommendation
if sentiment == "Positive":
    recommendation = "👍 Customer is satisfied with the product/service."

elif sentiment == "Negative":
    recommendation = "⚠️ Customer dissatisfaction detected. Improvement may be required."

else:
    recommendation = "ℹ️ Customer opinion is neutral."


# Display result
print("\n" + "=" * 55)
print("          AI-BASED ONLINE REVIEW ANALYZER")
print("=" * 55)

print("\nReview:")
print(review)

print("\nSentiment Analysis")
print("----------------------------")
print("Positive Score :", score["pos"])
print("Negative Score :", score["neg"])
print("Neutral Score  :", score["neu"])
print("Compound Score :", score["compound"])

print("\nFinal Sentiment:", sentiment)

print("\nReview Statistics")
print("----------------------------")
print("Word Count:", word_count)

print("\nImportant Keywords")
print("----------------------------")

if found_keywords:
    print(", ".join(found_keywords))
else:
    print("No specific keywords found.")

print("\nRecommendation")
print("----------------------------")
print(recommendation)

print("\n" + "=" * 55)
```
